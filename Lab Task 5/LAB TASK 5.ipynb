```python
g = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': [],
    'F': ['G'],
    'G': []
}

class MyStack:
    def __init__(self):
        self.a = []

    def add(self, x):
        self.a.append(x)

    def remove(self):
        if self.a:
            return self.a.pop()
        return None

def depth_first_search(s, t):
    v = []
    st = MyStack()
    st.add(s)

    while st.a:
        c = st.remove()
        if c not in v:
            v.append(c)
            if c == t:
                break
            for n in reversed(g[c]):
                st.add(n)
    return v

print(depth_first_search('A', 'F'))
```
