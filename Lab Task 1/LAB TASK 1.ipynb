```python
import random
def hangman():
    box = ["muzamil","abdullah","mehsam","Huraira","arman"]
    alphabet = random.choice(box).lower()
    rounds = 10
    Guessmade = ''
    entries = "abcdefghijklmnopqrstuvwxyz"
    while rounds >0:

        main_word = ""
        for i in alphabet:
            if i in Guessmade:
                main_word = main_word + i
            else:
                main_word = main_word + '_ '

        if main_word.replace(" ", "") == alphabet:
            print("You Won! The Game.")
            break
        print("Guess the word", main_word) 
        Guess = input("Enter a letter: ").lower()
        if Guess in entries and len(Guess) == 1:
            if Guess in Guessmade:
              print(" You already guessed that letter! ")
              continue
            else:
              Guessmade = Guessmade + Guess
        else:
            print("Enter Valid character")
            Guess = input().lower()
        if Guess not in alphabet:
            rounds =rounds - 1
            if rounds == 9:
                print("9 turns are left.!!")
                print("--------------------")
            if rounds == 8:
                print("8 turns are left.!!")
                print("--------------------")
                print("         O         ")
            if rounds == 7:
                print("7 turns are left.!!")
                print("--------------------")
                print("         O         ")
                print("         |         ")
            if rounds == 6:
                print("6 turns are left.!!")
                print("--------------------")
                print("         O         ")
                print("         |         ")
                print("        /          ")
            if rounds == 5:
                print("5 turns are left.!!")
                print("--------------------")
                print("         O         ")
                print("         |         ")
                print("        / \        ")

            if rounds == 4:
                print("4 turns are left.!!")
                print("--------------------")
                print("         O/         ")
                print("         |         ")
                print("        / \        ")

            if rounds == 3:
                print("3 turns are left.!!")
                print("--------------------")
                print("        \O/        ")
                print("         |         ")
                print("        / \        ")
            if rounds == 2:
                print("2 turns are left.!!")
                print("--------------------")
                print("        \O/  |     ")
                print("         |         ")
                print("        / \        ")
            if rounds == 1:
                print("1 turns are left.!!")
                print("--------------------")
                print("        \O/__|     ")
                print("         |         ")
                print("        / \        ")
            if rounds == 0:
                print("0 turns are left.!!")
                print("--------------------")
                print("        \O/__|     ")
                print("         |         ")
                print("        / \        ")
                print("-------      --------")
                print("You Lose .!!!")
                print("Game Over.")
                break



     

n = input("Enter the player name:")
print(f"Welcome {n}. in the hangman game.")
print("Guess the number in less than 10 attempts")
hangman()
```
