### 1. __Code for LUHN Algorithm__

```python
def luhn(n):
    n = n.replace(" ", "")
    s = 0
    r = n[::-1]

    for i in range(len(r)):
        d = int(r[i])
        if i % 2 == 1:
            d *= 2
            if d > 9:
                d -= 9
        s += d

    if s % 10 == 0:
        return "Valid Card Number"
    else:
        return "Invalid Card Number"
c = input("Enter card number: ")
print(luhn(c))
```

### **2. Write a python program to remove punctuations from the given string?**

```python
def remove(s):
    p = '''!()-[]{;}:'<>./?@,#$%^/&*_~'''
    o = ""
    for c in s:
        if c not in p:
            o += c
    return o

x = input("Enter something sentence:")
print(remove(x))
```

### **3. Write a python program to sort the sentence in alphabetical order?**

```python
def aligned(t):
    words = t.split()
    words.sort() 
    return " ".join(words)

t = input("Enter sentece u want to aligned")
print(aligned(t))
```
